##### Vraag 1a Het potentiaal rond een puntlading

1. Bepaal welke index correspondeert met $z = 0$ m.
2. Plot de log van het potentiaal tegen de log van de straal.
3. Fit de log van het potentiaal met de log van de straal.
4. Print het exponent op een nette manier.

````python
#1.
index_z0 = np.argmin(np.abs(z)) #zoeken naar het absolute minimum, dat zal dan ook het punt het dichtste bij z=0m zijn
print("Index dichtst bij z = 0m:",index_z0)


#2.
plt.figure()
plt.loglog(r[1:],V[index_z0,1:],'k.') #op log schaal de afstand met potentiaal plotten, eerste term r weghalen (want is 0 en log(0) bestaat niet), V alleen bij z=0 
# bijbehorende term V weghalen, (want anders een term te veel)
plt.xlabel('$r$ (m)')
plt.ylabel('$V$ (v)')
plt.show()

#3.
from scipy.optimize import curve_fit

def linear_fit(x,a,b): # het ziet er uit als een lineaire functie bij het loglog plot
    return a*x + b

val, cov = curve_fit(linear_fit, np.log(r[1:]), np.log(V[index_z0, 1:])) # curvefit op basis van de lineaire functie, het log nemen van de straal en potentiaal is heel belangrijk hier, 
# anders is het geen lineair verband.

a,b = val

r_fit = np.linspace(min(r[1:]), max(r[1:]),1000) #creeeren van een array van alle waardes van r voor de fitlijn
V_fit = np.exp(b) * r_fit**a # V berekenen uit a en b via ln(V) = ln(r) * a + b

plt.figure()
plt.loglog(r[1:], V[index_z0,1:], 'k.')
plt.loglog(r_fit, V_fit, 'r-') #plotten van de fitlijn
plt.xlabel('$r$ (m)')
plt.ylabel('$V$ (V)')
plt.show()

#4.
print("De exponent die uit de fit volgt is",a)
````

````python

def step(z):
    if z >= 0 :
        return sigma / (2*epsilon0)
    else :
        return -1* sigma / (2*epsilon0)
    # verwerken van het absolute-waarde-teken
E_theory_z = np.array([]) # maken van lege array, want direct array invullen gaf error
for i in z[1:-1] :
    E_theory_z = np.append(E_theory_z,step(i))
    # er wordt gelooped over de z-waarden, de theoretische E waarden worden berekend en worden vervolgens toegevoegd aan de array

plt.figure()
plt.title(f"Elektrisch veld langs de z-as voor een uniform geladen schijf")
plt.plot(z[1:-1], Ez[:,idx_r_zero], "o-", label="numeriek elektrisch veld") # plotten van z tegen de z component van het numeriek berekende elektrische veld
# z wordt gespliced zodat het evenveel waardes heeft als Ez
plt.plot(z[1:-1], E_theory_z, "r-",label="theoretisch elektrisch veld") # plotten van z tegen de z component van het theoretische elektrische veld
plt.legend()
plt.xlabel("$z$ (m)")
plt.ylabel("$E$ (V/m)")
plt.savefig("Elektrisch_veld_de_z-as_voor_een_uniform_geladen_schijf.png")
plt.show()
````

### Antwoord 3b.1

Voor een geleider geldt uit de randvoorwaarde van Maxwell:

$$E = \frac{\sigma}{\epsilon_0} 
\\\sigma = \epsilon_0 \cdot E$$

waar:

- $\sigma$ = oppervlakteladingsdichtheid (C/m$^2$)  
- $\varepsilon_0$ = permittiviteit van het vacuüm  
- $E$ = elektrische veld loodrecht op het oppervlak net buiten de geleider   



In de numerieke simulatie nemen we hiervoor het elektrische veld net boven het oppervlak van de schijf op de as.

````python
# we hebben E in het centrum al berekend hierboven

sigma_center = epsilon0 * E_center # met formule van hierboven 

print("\nOppervlakteladingsdichtheid in het midden van de schijf")
print(f"σ(r=0, z≈0) = {sigma_center:.4e} C/m²")
````

````python
# de waarden van de condenssatoropdracht woensdag zelf en de verbeterde( d is groter en epsilon karton is ook groter)
epsilon_0=8.854e-12 #F/m 
epsilon_r_woensdag=1
epsilon_karton_woensdag=epsilon_r_woensdag*epsilon_0 #epsilpon_r= ongveer 3 voor karton
d_woensdag= 2e-3 #afstand tussen de platen 
g=9.812 #m/s^2 valversnelling
k=100 #N/m veerconstante van onze veer geschat op 1 N/cm
L_0=14.184e-2 #begin lengte in cm
L=np.linspace(L_0,0.5*L_0) #lengte neemt af als resultaat van het gewicht dat er aan hangt van L_0 tot de helft van L_0
B=10e-2 #breede van de plaat blijf gelijk (10cm)
epsilon_r=2.3
epsilon_karton=epsilon_r*epsilon_0 #epsilpon_r= ongveer 3 voor karton
d= 4e-3 #afstand tussen de platen 
d_karton = 0.3e-3
d_lucht = 1.0e-3


# de functies die gebruikt worden
def massa(k,g,L_0,C,d,epsilon,B):
    #de massa van het object dat hangt(komt van k*x=m*g, met x=L_0-C*d/epsilon*B), x komt van de formule C=epsilon*B*(L_0-x)/d
    '''
    k=veerconstante in N/m
    g=valversnelling in m/s^2
    C=capaciteit in F
    d=afstand tussen platen in m
    epsilon=permitiviteit in F/m
    B = plaatbreedte in m
    '''
    return (k/g)*(L_0-C*d/(epsilon*B))


def vlakkeplaat(epsilon,L,B,d):
    '''
    epsilon = permitiviteit medium
    L = lengte in m
    B = breedte in m
    d = afstand tussen platen in m
    '''
    C = epsilon*B*L/d 
    return C

def capaciteit_met_karton(L,B,d_lucht,d_karton,epsilon_r):
    
    A = B*L
    epsilon0 = 8.854e-12
    
    C = A/(d_lucht/epsilon0 + d_karton/(epsilon0*epsilon_r))
    
    return C

C_min=vlakkeplaat(epsilon_karton_woensdag,0.5*L_0,B,d_woensdag)*1e12
C_max=vlakkeplaat(epsilon_karton_woensdag,L_0,B,d_woensdag)*1e12
C_schuif_woensdag=vlakkeplaat(epsilon_karton_woensdag,L,B,d_woensdag) #de capaciteit
C_nieuw = capaciteit_met_karton(L,B,d_lucht,d_karton,epsilon_r)
m=massa(k,g,L_0,C_schuif_woensdag,d,epsilon_karton,B)

plt.figure()
plt.plot(C_schuif_woensdag*1e12,m,label='originele schatting') # dit is de plot van de woensdag opdracht zelf
plt.ylabel("$massa$ (Kg)")
plt.xlabel("$Capaciteit$ (pF)")
plt.show()

print("min capaciteit is %f pF"%C_min)
print("max capaciteit is %f pF"%C_max)
print("max massa is op basis van simulatie %.3f kg"%max(m))

#3. 
print('De analytische oplossing van de capaciteit is',vlakkeplaat(epsilon_karton,L_0,B,d)*10**12,'pF')



plt.figure()
plt.plot(C_nieuw*1e12,m,label='gecorrigeerde schatting') # dit is de verbeterde plot
plt.ylabel("$massa$ (Kg)")
plt.xlabel("$Capaciteit$ (pF)")
plt.show()

C_min=vlakkeplaat(epsilon_karton,0.5*L_0,B,d)*1e12
C_max=vlakkeplaat(epsilon_karton,L_0,B,d)*1e12
print("min capaciteit is %f pF"%C_min)
print("max capaciteit is %f pF"%C_max)
print("max massa is op basis van simulatie %.3f kg"%max(m))
````
