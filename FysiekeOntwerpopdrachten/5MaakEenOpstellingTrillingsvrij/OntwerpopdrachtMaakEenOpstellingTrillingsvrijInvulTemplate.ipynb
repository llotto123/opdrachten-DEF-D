## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
W_0=np.sqrt((5*2*np.pi)**2/4)
f_kantel=W_0/(2*np.pi)
print(f_kantel)
```

```python
# Voeg hier je plot in.
def function(f,f_kantel):
    return np.abs(1/(1-(f**2/f_kantel**2)))

x_test=np.linspace(0,100,1000)
y_test=function(x_test,f_kantel)

plt.figure()
plt.loglog(x_test,y_test)
plt.xlabel('log(f)($Hz$)')
plt.ylabel('verhuding amplitude')
plt.ylim(0,10**3)
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
#W_0=np.sqrt(g/delta_x) dus:
g=9.81 #N/kg
delta_x=g/W_0**2 #m
print('De minimale delta x is ',delta_x)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

def functc(m,W_0):
    return W_0**2*m

m_test=np.linspace(0,10,1000)
C_test=functc(m_test,W_0)

plt.figure()
plt.plot(m_test,C_test,label='de berekende delta x')
plt.plot(m_test,(1/2)*C_test, label='factor 2 van delta x')
plt.plot(m_test,(1/5)*C_test, label='factor 5 van delta x')
plt.ylabel('C($N/m$)')
plt.xlabel('m($kg$)')
plt.legend()
plt.show()
print(functc(0.57,(1/5)*W_0))
```
